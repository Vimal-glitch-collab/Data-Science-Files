```python
x = df.drop("classification", axis=1)
x
```

```python
x.shape
```

```python
y = df["classification"]
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from keras.models import Sequential
from keras.layers import Dense

model= Sequential()

model.add(Dense(12, activation='relu', input_dim=24, kernel_initializer='uniform'))
model.add(Dense(8, activation='relu', kernel_initializer='uniform'))
model.add(Dense(1, activation='sigmoid', kernel_initializer='uniform'))
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(x_train, y_train, epochs=150, batch_size=16, verbose=1)
```

```python
acc_score = model.evaluate(x_test, y_test)
```

```python
model.metrics_names
```

```python
print(model.metrics_names[1],acc_score[1]*100)
```
