```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
df = pd.read_csv(r"C:\Users\vimal_z0\Downloads\ML Files\EDA\Loan_Data_Cleaned.csv")
df
```

```python
x = df.drop(["Loan_Status"], axis=1)
x
```

```python
y = df["Loan_Status"]
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
from keras.models import Sequential
from keras.layers import Dense

model = Sequential()

model.add(Dense(16, activation='relu',input_dim=11, kernel_initializer='uniform'))
model.add(Dense(8, activation='relu', kernel_initializer='uniform'))
model.add(Dense(1, activation='sigmoid', kernel_initializer='uniform'))
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=["accuracy"])
```

```python
model.fit(x_train, y_train, epochs=100, batch_size=16, verbose=2)
```

```python
model.metrics_names
```

```python
acc_score = model.evaluate(x_test, y_test)
```

```python
print(model.metrics[1], acc_score[1]*100)
```
