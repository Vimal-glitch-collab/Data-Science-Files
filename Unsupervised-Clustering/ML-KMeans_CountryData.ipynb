```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df["country"] = le.fit_transform(df["country"])
df
```

```python
from sklearn.preprocessing import MinMaxScaler
minmax = MinMaxScaler(feature_range=(0,1))
X = minmax.fit_transform(df)
X1 = pd.DataFrame(X, columns=['country',	'child_mort','exports',	'health',	'imports',	'income',	'inflation',	'life_expec',	'total_fer',	'gdpp'])
X1.describe()
```

```python
df1 = pd.read_csv(r"C:\Users\vimal_z0\Downloads\ML Files\EDA\CountryData_Cleaned.csv")
df1
```

```python
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,11):
    kmn = KMeans(n_clusters=i, init= "k-means++", random_state=42)
    model = kmn.fit_predict(df1)
    wcss.append(kmn.inertia_)
```

```python
plt.plot(wcss)
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
```

```python
from sklearn.cluster import KMeans
kmn = KMeans(n_clusters=7, init= "k-means++", random_state=42)
model = kmn.fit_predict(df1)
```

```python
from sklearn.metrics import silhouette_score
sil_score = silhouette_score(df1, model)
print("Silhouette Score:", sil_score)
```
