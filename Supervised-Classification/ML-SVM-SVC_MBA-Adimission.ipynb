```python
for i in freqgraph_flt:
    df[i] = df[i].fillna(df[i].median())
df.isnull().sum()
```

```python
freq_obj = df.select_dtypes(include="str").columns
freq_obj
```

```python
for i in freq_obj:
    df[i] = df[i].fillna(df[i].mode()[0])
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
for i in freq_obj:
    df[i] = le.fit_transform(df[i])
df
```

```python
X = df.drop("Specialization", axis =1)
X
```

```python
Y = df["Specialization"]
Y
```

```python
from sklearn.preprocessing import MinMaxScaler
minmax = MinMaxScaler()
X1 = minmax.fit_transform(X)
X1 = pd.DataFrame(X1, columns= X.columns)
X1 = pd.concat([X1,Y], axis =1)
X1.describe()
```

```python
df1= pd.read_csv(r"C:\Users\vimal_z0\Downloads\ML Files\EDA\MBA-Adimission_Cleaned.csv")
df1
```

```python
x = df1.drop("Specialization", axis =1)
x
```

```python
y = df["Specialization"]
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size =0.25, random_state=42)
```

```python
from sklearn.svm import SVC
svm_cln = SVC(kernel="rbf")
model = svm_cln.fit(x_train, y_train)
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, recall_score, f1_score, precision_score, classification_report
print("confusion_matrix:", confusion_matrix(y_pred, y_test))
print("accuracy score:", accuracy_score(y_test, y_pred))
print("recall score:", recall_score(y_pred, y_test,average ="macro"))
print("f1_score:", f1_score(y_pred, y_test,average ="macro"))
print("precision score:", precision_score(y_pred, y_test,average ="macro"))
print(classification_report(y_pred, y_test))
```
