```python
import seaborn as sns    
import pandas as pd
import matplotlib.pyplot as plt                         
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\vimal_z0\Downloads\ML Files\EDA\Heart_Data_Cleaned.csv")
df
```

```python
x = df.drop("HeartDisease", axis=1)
x
```

```python
y = df["HeartDisease"]
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.25, random_state=42)
```

```python
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB()
model1 = gnb.fit(x_train, y_train)
y_pred1 = model1.predict(x_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
mnb = MultinomialNB()
model2 = mnb.fit(x_train, y_train)
y_pred2 = model2.predict(x_test)
```

```python
from sklearn.naive_bayes import BernoulliNB
bnb = BernoulliNB()
model3 = bnb.fit(x_train, y_train)
y_pred3 = model3.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
print("Accuracy of Model1:", accuracy_score(y_pred1, y_test))
print("Accuracy of Model2:", accuracy_score(y_pred2, y_test))
print("Accuracy of Model3:", accuracy_score(y_pred3, y_test))
```
