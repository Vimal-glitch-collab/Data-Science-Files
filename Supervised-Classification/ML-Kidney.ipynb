```python
df1 = pd.read_csv(r"Kidney_data.csv")
df1
```

```python
x = df1.drop(["classification"], axis =1)
x
```

```python
y = df1["classification"]
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state =42)
```

```python
from sklearn.linear_model import LogisticRegression
logreg =LogisticRegression(max_iter = 100)
```

```python
model = logreg.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score, precision_score, recall_score, classification_report
print("Confusion Matrix:", confusion_matrix(y_pred, y_test))
print("Accuracy Score:", accuracy_score(y_pred, y_test))
print("F1 Score:", f1_score(y_pred, y_test,average="macro"))
print("Precision Score:", precision_score(y_pred, y_test, average ="macro"))  #average = "macro" is used for multi-class classification to calculate metrics for each label, and find their unweighted mean. This does not take label imbalance into account.
print("Recall Score:", recall_score(y_pred, y_test, average ="macro"))
print(classification_report(y_pred, y_test))
```
