```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\vimal_z0\Downloads\ML Files\EDA\Loan_Data_Cleaned.csv")
df
```

```python
x = df.drop("Loan_Status", axis =1)
x
```

```python
y = df["Loan_Status"]
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.25, random_state=42)
```

```python
from sklearn.svm import SVC  #Support Vector Classifier
svm_clf = SVC(kernel="rbf") #kernels used-> rbf(Radial basis function), linear and poly. it is used to create margin between classes for output like True/False, Yes/No, etc....
model = svm_clf.fit(x_train, y_train)
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score, recall_score, precision_score, classification_report
```

```python
print(confusion_matrix(y_test, y_pred))
```

```python
print(accuracy_score(y_test, y_pred))
```

```python
print(f1_score(y_test, y_pred))
```

```python
print(recall_score(y_test, y_pred))
```

```python
print(precision_score(y_pred, y_test))
```

```python
print(classification_report(y_test, y_pred))
```
