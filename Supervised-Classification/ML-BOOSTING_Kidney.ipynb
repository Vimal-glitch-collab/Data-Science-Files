```python
df = pd.read_csv(r"C:\Users\vimal_z0\Downloads\ML Files\EDA\Kidney_data.csv")
df
```

```python
x = df.drop("classification", axis=1)
y = df["classification"]
print(x)
print(y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.25, random_state=42)
```

```python
from sklearn.ensemble import AdaBoostClassifier
adb = AdaBoostClassifier()
model1 = adb.fit(x_train, y_train)
y_pred1 = model1.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score, recall_score, precision_score, f1_score, classification_report
print("confusion matrix of AdaBoost:",confusion_matrix(y_pred1,y_test))
print("Accuracy Score of AdaBoost:", accuracy_score(y_pred1, y_test))
print("Recall score of AdaBoost:", recall_score(y_pred1, y_test, average="macro"))
print("Precision Score of AdaBoost:", precision_score(y_pred1, y_test,average="macro"))
print("f1 score of AdaBoost:", f1_score(y_pred1, y_test,average="macro"))
print("classification report of AdaBoost:", classification_report(y_pred1, y_test))
```

```python
from sklearn.ensemble import GradientBoostingClassifier
gbc = GradientBoostingClassifier()
model2= gbc.fit(x_train, y_train)
y_pred2= model2.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score, recall_score, precision_score, f1_score, classification_report
print("confusion matrix of GradientBoosting:",confusion_matrix(y_pred2,y_test))
print("Accuracy Score of GradientBoosting:", accuracy_score(y_pred2, y_test))
print("Recall score of GradientBoosting:", recall_score(y_pred2, y_test, average="macro"))
print("Precision Score of GradientBoosting:", precision_score(y_pred2, y_test,average="macro"))
print("f1 score of GradientBoosting:", f1_score(y_pred2, y_test,average="macro"))
print("classification report of GradientBoosting:", classification_report(y_pred2, y_test))
```

```python
from xgboost import XGBClassifier
xgb = XGBClassifier()
model3 = xgb.fit(x_train, y_train)
y_pred3 = model3.predict(x_test) 
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score, recall_score, precision_score, f1_score, classification_report
print("confusion matrix of XGBoosting:",confusion_matrix(y_pred3,y_test))
print("Accuracy Score of XGBoosting:", accuracy_score(y_pred3, y_test))
print("Recall score of XGBoosting:", recall_score(y_pred3, y_test, average="macro"))
print("Precision Score of XGBoosting:", precision_score(y_pred3, y_test,average="macro"))
print("f1 score of XGBoosting:", f1_score(y_pred3, y_test,average="macro"))
print("classification report of XGBoosting:", classification_report(y_pred3, y_test))
```
