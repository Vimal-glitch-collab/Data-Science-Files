Preprocessing Text Data

```python
def wordopt(text):
    text = text.lower()
    text = re.sub('\[.*?\]','',text)
    text = re.sub("\\W"," ",text)
    text = re.sub('https?://\S+|www\.\S+','',text)
    text = re.sub('<.*?>+','',text)
    text = re.sub('[%s]'% re.escape(string.punctuation),'',text)
    text = re.sub('\n','',text)
    text = re.sub('\w*\d\w*','',text)
    return text    
```

```python
df["text"] = df["text"].apply(wordopt)
```

Data Visualization

```python
x = df["text"]
y = df["class"]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.25, random_state=42)
```

```python
vectorization = TfidfVectorizer()
xv_train = vectorization.fit_transform(x_train)
xv_test = vectorization.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
lg = LogisticRegression(max_iter=100)
lg.fit(xv_train, y_train)
pred_lg = lg.predict(xv_test)
print(pred_lg)
```

```python
print(xv_test)
```

```python
lg_acc = accuracy_score(pred_lg,y_test)
print(lg_acc)
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc = DecisionTreeClassifier()
dtc.fit(xv_train, y_train)
```

```python
dtc_acc = dtc.score(xv_test, y_test)
print(dtc_acc)
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier()
rfc.fit(xv_train,y_train)
```

```python
rfc_acc= rfc.score(xv_test,y_test)
print(rfc_acc)
```

```python
performance =[]
for i in [lg, dtc, rfc]:
    i.fit(xv_train, y_train)
    performance.append(round(accuracy_score(y_test, i.predict(xv_test))*100,2))
print(performance)
```
