```python
import re
import string

def wordopt(text):
    text = text.lower()
    text = re.sub('\[.*?\]','',text)
    text = re.sub("\\W"," ",text)
    text = re.sub('https?://\S+|www\.\S+','',text)
    text = re.sub('<.*?>+','',text)
    text = re.sub('[%s]'%re.escape(string.punctuation),'',text)
    text = re.sub('\n','',text)
    text = re.sub('\w*\d\w*','',text)
    return text
```

```python
df['text'] = df['text'].apply(wordopt)
```

```python
from wordcloud import WordCloud

txt =" ".join([i for i in df["text"]])
plt.figure(figsize=(20,10), facecolor=None)
wordcloud = WordCloud(max_words=500, width=1600, height=1000).generate(txt)
plt.imshow(wordcloud, interpolation='bilinear')
plt.title("Word cloud of Twitter data")
plt.axis("off")
plt.show()
```

```python
x = df["text"]
y = df["label"]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
xv_train=vectorizer.fit_transform(x_train)
xv_test = vectorizer.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
lr = LogisticRegression(max_iter=100)
lr.fit(xv_train, y_train)
lr_acc = accuracy_score(y_test, lr.predict(xv_test))
print("Accuracy of Logistic:", round(lr_acc*100,2))
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc = DecisionTreeClassifier(random_state=42)
dtc.fit(xv_train, y_train)
dtc_acc = dtc.score(xv_test, y_test)
print("Accuracy Score of Decision Tree:", round(dtc_acc*100,2))
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier(random_state=42)
rfc.fit(xv_train, y_train)
rfc_acc = rfc.score(xv_test, y_test)
print("Accuracy Score of Random forest:", round(rfc_acc*100,2))
```
