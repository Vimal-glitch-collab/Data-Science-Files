```python
med = df.select_dtypes(include = ["float"])
med1 = med.columns.drop("Rating")
med1
```

```python
for i in med1:
    df[i] = df[i].fillna(df[i].median())
df.isnull().sum()
```

```python
df["Rating"] = df['Rating'].fillna(df["Rating"].mean())
df.isnull().sum()
```

```python
mod = df.select_dtypes(["object", "str"]).columns
mod
```

```python
for i in mod:
    df[i] = df[i].fillna(df[i].mode()[0])
df.isnull().sum()

    
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df.drop(["index"], axis =1, inplace =True)
df.drop(["Model"],axis =1, inplace = True)
```

```python
df1 = pd.read_csv(r"C:\Users\vimal_z0\Downloads\ML Files\Laptops_Cleaned.csv")
df1
```

```python
A = df1.drop(["Price"], axis =1)
A
```

```python
B = df1["Price"]
B
```

```python
from sklearn.model_selection import train_test_split
A_train, A_test, B_train, B_test = train_test_split(A,B,test_size =0.25,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
model = lr.fit(A_train, B_train)
```

```python
B_pred = model.predict(A_test)
B_pred
```

```python
from sklearn.metrics import mean_squared_error
print("MSE :",mean_squared_error(B_test, B_pred))
```

```python
from sklearn.metrics import r2_score
print("r2_score :", r2_score(B_pred, B_test))
```
