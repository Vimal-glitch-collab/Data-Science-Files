```python
import numpy as np          
import matplotlib.pyplot as plt                      
import seaborn as sns  
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\vimal_z0\Downloads\ML Files\EDA\HousePricing_Cleaned.csv")
df
```

```python
x = df.drop("Price", axis=1)
y = df["Price"]
print(x)
print(y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.25, random_state=42)
```

```python
from sklearn.ensemble import AdaBoostRegressor
abc = AdaBoostRegressor()
model1 = abc.fit(x_train, y_train)
y_pred1 = model1.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
print("MSE of AdaBoost:", mean_squared_error(y_pred1, y_test))
print("r2 score of AdaBoost:", r2_score(y_pred1, y_test))
```

```python
from sklearn.ensemble import GradientBoostingRegressor
gbc = GradientBoostingRegressor()
model2 = gbc.fit(x_train, y_train)
y_pred2 = model2.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
print("MSE of Gradient Boosting:", mean_squared_error(y_pred2, y_test))
print("r2 score of Gradient Boosting:", r2_score(y_pred2, y_test))
```

```python
from xgboost import XGBRegressor
xgb = XGBRegressor()
model3 = xgb.fit(x_train, y_train)
y_pred3 = model3.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
print("MSE of XGBoosting:", mean_squared_error(y_pred3, y_test))
print("r2 score of XGBoosting:", r2_score(y_pred3, y_test))
```
