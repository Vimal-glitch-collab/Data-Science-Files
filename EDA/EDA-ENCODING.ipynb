fit_transform() -> is performed as label encoding where fit reads the data and transform converts the categorical(object) data in numerical values

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df["country"] = le.fit_transform(df["country"])  #fit()+transform() = fit_transform() , fit() looks at the unique categories in your column and learns the categories.
df["continent"] = le.fit_transform(df["continent"])  #transform() converts the learned categories into numbers.  eg: Asia =0, europe =1, etc...
df
```

```python
df.drop("country", axis =1, inplace = True)  # as country column has unique values, it will not be useful for prediction, because it is same as index values.So we can drop it.
```
