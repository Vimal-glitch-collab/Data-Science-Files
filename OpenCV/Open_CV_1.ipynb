```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from skimage import io   #for image reading
from PIL import Image
import os
```

```python
img = cv2.imread(r"C:\Users\vimal_z0\Downloads\1.jpeg")
plt.imshow(img)
```

```python
converted_image = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  #convert BGR to RGB for displaying with matplotlib

plt.imshow(converted_image)
plt.axis('off')
plt.show()
```

```python
resized_img = cv2.resize(converted_image,(300, 300))
plt.imshow(resized_img)
plt.axis('off')
```

```python
flip_v = cv2.flip(resized_img, 0)  #vertical -> upside down
plt.imshow(flip_v)
```

```python
flip_h = cv2.flip(resized_img, 1) #horizontal -> left to right
plt.imshow(flip_h)
```

```python
flip_b = cv2.flip(resized_img, -1) #both vertical and horizontal
plt.imshow(flip_b)
```

```python
(h, w) = resized_img.shape[:2] #0,1,2  0-height, 1- width, 2-channel
center = (w // 2, h // 2)

matrix = cv2.getRotationMatrix2D(center, 45, 1.0) #center, angle, scale
rotated_img = cv2.warpAffine(resized_img, matrix, (w, h))
plt.imshow(rotated_img)
plt.axis('off')
```

```python
#used for blurring the image and noise removal. It is a low pass filter which removes the high frequency noise from the image.
blur_img = cv2.GaussianBlur(converted_image, (3,3), 0)   #3,3 or 5,5
plt.imshow(blur_img)
plt.axis('off')
```

```python
edges = cv2.Canny(converted_image,100,200)
plt.imshow(edges, cmap='gray')
plt.axis('off')
```

```python
cir_img = cv2.circle(converted_image, center=(400,100), radius=60, color=(255,0,100), thickness=4)
plt.imshow(cir_img)
```
