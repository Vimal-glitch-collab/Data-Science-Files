```python
import cv2
import matplotlib.pyplot as plt
```

```python
face_cascade = cv2.CascadeClassifier(r"C:\Users\vimal_z0\Downloads\ML Files\OpenCV\haarcascade_frontalface_default.xml")
```

```python
# vid = cv2.VideoCapture(r"C:\Users\vimal_z0\Downloads\WhatsApp Video 2026-11-04 at 13.11.20.mp4")
vid = cv2.VideoCapture(0)
while True: #Creates an infinite loop to read video frames continuously
    success, frame = vid.read()
# Reads one frame at a time from the video.
# success → True if frame read correctly
# frame → the actual image (numpy array)    
    if frame is not None and frame.shape[0] > 0 and frame.shape[1] > 0:
        # Checks if the frame is valid (not empty)
        faces = face_cascade.detectMultiScale(frame)
        # detectMultiScale() returns multiple rectangles around detected faces
        for (x,y,w,h) in faces:
            cv2.rectangle(frame,(x,y),(x+w,y+h),color=(255,0,255),thickness=3)     
            cv2.imshow("Video", frame)

    if cv2.waitKey(1) & 0xFF == 27:
        break
vid.release()
cv2.destroyAllWindows()
```
