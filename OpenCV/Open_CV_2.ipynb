```python
import cv2
from skimage import io
from PIL import Image 
import matplotlib.pyplot as plt    
import pandas as pd
import numpy as np
```

```python
img = cv2.imread(r"C:\Users\vimal_z0\Downloads\gettyimages-605559123-612x612.jpg")
plt.imshow(img)
plt.axis('off')
```

```python
cnt_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(cnt_img)
plt.axis('off')
plt.show()
```

```python
gray = cv2.cvtColor(cnt_img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
```

```python
# !python -m pip install opencv-python==4.10.0.84
```

```python
face_cascade = cv2.CascadeClassifier(r"C:\Users\vimal_z0\Downloads\ML Files\OpenCV\haarcascade_frontalface_default.xml")
```

```python
faces = face_cascade.detectMultiScale(gray, scaleFactor=1.3, minNeighbors=5)
```

```python
print(len(faces))
```

```python
for (x,y,w,h) in faces:
    cv2.rectangle(gray, (x,y), (x+w, y+h), (255, 255, 0), 5)
    
plt.imshow(gray, cmap ='gray')
plt.axis('off')
```
