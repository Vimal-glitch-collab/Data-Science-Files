```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\Lenovo\Downloads\img-1.jpeg")
```

```python
converted_image = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(converted_image)
plt.axis("off")
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

#to read harcascade file(harcascade file is algorithm for detect object and face),false positive and error increase.because it is less trained
#(only used for study)


```python
face_cascade = cv2.CascadeClassifier(
    r"C:\Users\Lenovo\Downloads\haarcascade_frontalface_default.xml"
)
```

```python
# face_cascade.load('haarcascade_frontalface_default .xml')  # if you have any error use this code
```

```python
faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.3,
    minNeighbors=5
) 

# A trained Haar Cascade classifier used for face detection.
# The grayscale image-Face detection usually works better on grayscale imagesscaleFactor=1.3

# scaleFactor=1.3
# Specifies how much the image size is reduced at each scale.
# 1.3 means the image is reduced by 30% each time.

# minNeighbors=5-Defines how many nearby detections are required to keep a detected face
```

```python
#to find faces from the image

print(len(faces))
```

```python
# (x, y): Top-left corner of the rectangle.
# (x + w, y + h): Bottom-right corner of the rectangle.
# color=(255, 0, 255): The color of the rectangle in BGR format (magenta).
# thickness=3: The thickness of the rectangle’s border.
for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x+w, y+h), (0,255,0), 2)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
```
